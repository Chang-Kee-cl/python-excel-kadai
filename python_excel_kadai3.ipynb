```python
import pandas as pd

import openpyxl

file_list = ['2022_年間売上表.xlsx', '2023_年間売上表.xlsx']

df = pd.DataFrame()

for file_name in file_list:
    temp_df = pd.read_excel(file_name)
    temp_df.columns = ['売上年', '商品', '金額（千円）']
    df = pd.concat([df, temp_df], ignore_index=True)
```

```python
grouped = df.groupby(['商品','売上年'])['金額（千円）'].sum()
# df = df.sort_values('商品', ascending=True)
```

```python
writer = pd.ExcelWriter('売上集計表.xlsx')

grouped.to_excel('売上集計表.xlsx')
```

```python
from openpyxl.styles import PatternFill

workbook = openpyxl.load_workbook('売上集計表.xlsx')
sheet = workbook['Sheet1']

for raw in sheet['A1:C1']:
    for cell in raw:
        cell.fill = PatternFill(patternType='solid', fgColor='F2F2F2')

workbook.save('売上集計表.xlsx')
```
